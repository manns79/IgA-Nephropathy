# Trim Lab Data
The objective of this notebook is to trim *kidney_disease_random_1000_patients.csv.gz* and *normal_kidney_patients.csv.gz* by reducing the types of labs included to those that are specified in *Minimal Reducedlabitems - Reducedlabitems.csv*. This yields a smaller data set that still contains all of the information that is relevant to our study. We start by trimming *kidney_disease_random_1000_patients.csv.gz*.

```python
import pandas as pd
import numpy as np
import json
```

```python
kidney_disease_data = pd.read_csv('../Data/kidney_disease_random_1000_patients.csv.gz',
                                 compression = 'gzip')
kidney_disease_data.head()
```

```python
reduced_labs = pd.read_csv('../Data/Minimal Reducedlabitems - Reducedlabitems.csv')
reduced_labs.head()
```

```python
reduced_labs['itemid'].shape[0]
```

```python
reduced_ids = list(reduced_labs['itemid'])
labevents_cols = ['labevent_id',
                 'subject_id',
                 'hadm_id',
                 'specimen_id',
                 'itemid',
                 'order_provider_id',
                 'charttime',
                 'storetime',
                 'value',
                 'valuenum',
                 'valueuom',
                 'ref_range_lower',
                 'ref_range_upper',
                 'flag',
                 'priority',
                 'comments']
trimmed_kidney_disease_data = pd.DataFrame(columns = labevents_cols)
for lab in reduced_ids:
    lab_df = kidney_disease_data.loc[kidney_disease_data['itemid'] == lab]
    trimmed_kidney_disease_data = pd.concat([trimmed_kidney_disease_data, lab_df])
```

```python
trimmed_kidney_disease_data.to_csv('../Data/trimmed_kidney_disease_random_1000_patients.csv.gz',
                                  compression = 'gzip',
                                  index = False)
print('The number of lab items in the original data set is', kidney_disease_data['itemid'].value_counts().shape[0], '\n')
print('The number of lab items in the new data set is', trimmed_kidney_disease_data['itemid'].value_counts().shape[0])
```

As may be seen from the output of the cell above, the original data set included 826 lab items, while the trimmed data set only includes 95. The remainder of this notebook trims *normal_kidney_patients.csv.gz* in the same manner. 

```python
normal_kidney_data = pd.read_csv('../Data/normal_kidney_patients.csv.gz',
                                compression = 'gzip')
normal_kidney_data.head()
```

```python
trimmed_normal_kidney_data = pd.DataFrame(columns = labevents_cols)
for lab in reduced_ids:
    lab_df = normal_kidney_data.loc[normal_kidney_data['itemid'] == lab]
    trimmed_normal_kidney_data = pd.concat([trimmed_normal_kidney_data, lab_df])
```

```python
trimmed_normal_kidney_data.to_csv('../Data/trimmed_normal_kidney_patients.csv.gz',
                                 compression = 'gzip',
                                 index = False)
print('The number of lab items in the original data set is', normal_kidney_data['itemid'].value_counts().shape[0], '\n')
print('The number of lab items in the new data set is', trimmed_normal_kidney_data['itemid'].value_counts().shape[0])
```

Based on the output above, *normal_kidney_patients.csv.gz* has been trimmed from $693$ lab items to $76$ lab items.
