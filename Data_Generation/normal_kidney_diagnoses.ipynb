# Normal Kidney Diagnoses
This notebook is analogous to *kidney_disease_diagnoses.ipynb*, but pertains to patients with healthy kidneys. In particular, the objective is to create a dictionary called `normal_kidney_patient_diagnoses`, where the keys are the `subject_id` of a patient with healthy kidneys, and the values are tuples of the form `(seq_num, icd_code, icd_version)`. As in *kidney_disease_diagnoses.ipynb*, each entry in the tuple is a list whose length is determined by the number of diagnoses that the patient received. The `long_title` of the diagnosis can be obtained by using the `icd_code` in conjunction with *d_icd_diagnoses.csv*.

```python
import numpy as np
import pandas as pd
import json
```

```python
normal_kidney_data = pd.read_csv('../Data/normal_kidney_patients.csv.gz',
                                compression = 'gzip')
normal_kidney_data.head()
```

```python
diagnoses_data = pd.read_csv('../Data/diagnoses_icd.csv')
diagnoses_data.head()
```

```python
subject_ids = list(set(normal_kidney_data['subject_id']))
normal_kidney_patient_diagnoses = {}
for patient in subject_ids:
    patient_diagnoses = diagnoses_data.loc[diagnoses_data['subject_id'] == patient]
    normal_kidney_patient_diagnoses[patient] = (list(patient_diagnoses['seq_num']),
                                               list(patient_diagnoses['icd_code']),
                                               list(patient_diagnoses['icd_version']))
```

```python
with open('../Data/normal_kidney_patient_diagnoses.json', 'w') as file:
    json.dump(normal_kidney_patient_diagnoses, file)
```
