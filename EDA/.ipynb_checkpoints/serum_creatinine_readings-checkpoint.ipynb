# Serum Creatinine Readings
The objective of this notebook is to identify what proportion of the patients in the MIMIC-IV data have a serum creatinine reading conditional on the patient having received a kidney disease diagnosis. During the data generation process, we determined the `subject_id` values associated with patients that received a kidney disease diagnosis. Furthermore, we have reduced `labevents.csv` from MIMIC-IV to a file called `serum_creatinine_labs.csv` that contains all lab measurements of serum creatinine. After obtaining the `subject_id` values from `serum_creatinine_labs.csv`, the objective of this notebook can be accomplished by looking at the intersection of the two sets of `subject_id` values (i.e. patients with kidney disease and patients with a serum creatinine reading).

```python
import pandas as pd
import numpy as np
import json
```

```python
with open('../Data/all_subject_ids.json', 'r') as file:
    kidney_disease_ids = json.load(file)
kidney_disease_ids[:10]
```

```python
serum_creatinine_labs_df = pd.read_csv('../Data/serum_creatinine_labs.csv')
serum_creatinine_labs_df.head()
```

```python
serum_creatinine_ids = list(set(serum_creatinine_labs_df['subject_id']))
intersection_ids = set(serum_creatinine_ids).intersection(set(kidney_disease_ids))
print('The number of patients that were diagnosed with kidney disease is', len(kidney_disease_ids), '\n')
print('The number of patients with a serum creatinine reading is', len(serum_creatinine_ids), '\n')
print('The number of patients that were both diagnosed with kidney disease and have a serum creatinine reading is', len(intersection_ids), '\n')
print('The proportion of patients with kidney disease that have a serum creatinine reading is', np.round(len(intersection_ids)/len(kidney_disease_ids), 4))
```
