```python
import pandas as pd
import numpy as np
import json
```

```python
normal_kidney_df = pd.read_csv('../Data/modified_normal_kidney_patients.csv')
normal_kidney_df.head()
```

```python
d_icd_df = pd.read_csv('../Data/d_icd_diagnoses.csv')
d_icd_df.head()
```

```python
diagnoses_df = pd.read_csv('../Data/diagnoses_icd.csv')
diagnoses_df.head()
```

```python
normal_kidney_ids = list(set(normal_kidney_df['subject_id']))
normal_kidney_patients_dict = {}
for patient in normal_kidney_ids:
    patient_diagnoses_df = diagnoses_df.loc[diagnoses_df['subject_id'] == patient, ['seq_num', 'icd_code', 'icd_version']]
    normal_kidney_patients_dict[patient] = (list(patient_diagnoses_df['seq_num']),
                                           list(patient_diagnoses_df['icd_code']),
                                           list(patient_diagnoses_df['icd_version']))
    
```

```python
with open('../Data/normal_kidney_patients.json', 'w') as file:
    json.dump(normal_kidney_patients_dict, file)
```

```python
with open('../Data/normal_kidney_patients.json', 'r') as file:
    normal_kidney_dict = json.load(file)
```

```python
rand_int = np.random.randint(low = 0,
                            high = len(normal_kidney_ids))
subject_id = normal_kidney_ids[rand_int]
patient = normal_kidney_dict[f'{subject_id}']
icd_codes = patient[1]
print(f'Diagnoses for patient {subject_id}:')
for n, code in enumerate(icd_codes):
    diagnosis = d_icd_df.loc[d_icd_df['icd_code'] == code, 'long_title'].values[0]
    print(f'\t Diagnosis {n+1} is: {diagnosis}')
```
