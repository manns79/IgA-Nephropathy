# Modeling
The objective of this notebook is to apply different classification algorithms to our data. Requiring patients in our dataset to have several different lab readings taken on the same day significantly reduced the size of the dataset (764 patients), which makes certain classification algorithms unsuitable. With that in mind, we limit our study to the following algorithms:
1. LogisticRegression
2. KNeighborsClassifier

For both of these algorithms, we use GridSearchCV to perform 5-fold cross-validation. Furthermore, different values of `C` and `n_neighbors` are tried for LogisticRegression and KNeighborsClassifier, respectively, to optimize these hyperparameters. After doing cross-validation on the training set, our best model is evaluated on the test set. For the purpose of comparing with the paper [Chronic kidney disease prediction based on machine learning algorithms](https://doi.org/10.1016/j.jpi.2023.100189), we also perform a grid search using GradientBoostingClassifier on the training set. However, note that we are **not** advocating using a classification algorithm such as GradientBoostingClassifier on datasets that are as limited in size as the one we are working with. Rather, the purpose of this test is to illustrate that the results which we obtain are likely not as different as the results reported in the previously mentioned paper may appear at face value. 


```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import accuracy_score, confusion_matrix 
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier 
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.preprocessing import StandardScaler
sns.set_style('whitegrid')
```

```python
data = pd.read_csv('../Data/modeling_additional_labs.csv')
data_train, data_test = train_test_split(data,
                                        test_size = 0.2,
                                        random_state = 127,
                                        stratify = data['class'])
data_train.to_csv('../Data/data_train_unscaled.csv')
data_test.to_csv('../Data/data_test_unscaled.csv')
numerical_features = ['albumin',
                     'bicarbonate',
                     'calcium',
                     'chloride',
                     'creatinine',
                     'potassium',
                     'bun',
                     'hemoglobin',
                     'bun_over_creatinine']
scaler = StandardScaler().set_output(transform = 'pandas')
scaler.fit(data_train[numerical_features])
data_train[numerical_features] = scaler.transform(data_train[numerical_features])
data_test[numerical_features] = scaler.transform(data_test[numerical_features])
data_train.to_csv('../Data/data_train_scaled.csv')
data_test.to_csv('../Data/data_test_scaled.csv')
target = 'class'
data_train.head()
```

### Features: albumin

```python
features = ['albumin']
log_grid_cv = GridSearchCV(estimator = LogisticRegression(),
                          param_grid = {'C': [0.001, 0.01, 0.1, 1, 10, 100]},
                          scoring = 'accuracy',
                          cv = 5)
knn_grid_cv = GridSearchCV(estimator = KNeighborsClassifier(),
                          param_grid = {'n_neighbors': np.arange(1, 21, 1)},
                          scoring = 'accuracy',
                          cv = 5)
log_grid_cv.fit(data_train[features],
               data_train[target])
knn_grid_cv.fit(data_train[features],
               data_train[target])
log_best = log_grid_cv.best_estimator_
knn_best = knn_grid_cv.best_estimator_ 
log_cm = confusion_matrix(data_train[target],
                         log_best.predict(data_train[features]))
knn_cm = confusion_matrix(data_train[target],
                         knn_best.predict(data_train[features]))
print('The best value of C for logistic regression is', log_grid_cv.best_params_)
print('The best accuracy score for logistic regression is', log_grid_cv.best_score_)
print('The false negative rate for the logistic regression model with the highest accuracy is', log_cm[1, 0]/(log_cm[1, 1] + log_cm[1, 0]))
print('\n')
print('The best value of k for k-nearest neighbors is', knn_grid_cv.best_params_)
print('The best accuracy score for k-nearest neighbors is', knn_grid_cv.best_score_)
print('The false negative rate for the k-nearest neighbors model with the highest accuracy is', knn_cm[1, 0]/(knn_cm[1, 1] + knn_cm[1, 0]))
```
